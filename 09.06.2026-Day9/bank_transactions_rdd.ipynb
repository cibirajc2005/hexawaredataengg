```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("BankTransactions").getOrCreate()
sc = spark.sparkContext

transactions = [
    "DEPOSIT","WITHDRAW","TRANSFER","DEPOSIT","BALANCE_CHECK",
    "TRANSFER","DEPOSIT","WITHDRAW","LOAN_PAYMENT","DEPOSIT",
    "TRANSFER","WITHDRAW","DEPOSIT","BALANCE_CHECK","TRANSFER",
    "DEPOSIT","WITHDRAW","LOAN_PAYMENT","TRANSFER","DEPOSIT",
    "WITHDRAW","DEPOSIT","TRANSFER","BALANCE_CHECK","WITHDRAW",
    "DEPOSIT","TRANSFER","LOAN_PAYMENT","DEPOSIT","WITHDRAW",
    "TRANSFER","DEPOSIT","BALANCE_CHECK","TRANSFER","WITHDRAW",
    "DEPOSIT","TRANSFER","LOAN_PAYMENT","DEPOSIT","WITHDRAW",
    "TRANSFER","BALANCE_CHECK","DEPOSIT","TRANSFER","WITHDRAW",
    "LOAN_PAYMENT","DEPOSIT","TRANSFER","BALANCE_CHECK","DEPOSIT"
]
rdd = sc.parallelize(transactions)
print("RDD created successfully")
```

## Union and Intersection

```python
extra_transactions = ["DEPOSIT","ATM_WITHDRAWAL","TRANSFER","MOBILE_RECHARGE","BALANCE_CHECK"]
rdd2 = sc.parallelize(extra_transactions)
print("41. Union:", rdd.union(rdd2).collect())
```

```python
print("42. Count after union:", rdd.union(rdd2).count())
```

```python
print("43. Distinct after union:", rdd.union(rdd2).distinct().collect())
```

```python
print("44. New types introduced:", rdd2.subtract(rdd).distinct().collect())
```

```python
union_freq = rdd.union(rdd2).map(lambda x: (x,1)).reduceByKey(lambda a,b: a+b)
print("45. Frequencies after union:", union_freq.collect())
```

```python
rdd3 = sc.parallelize(["DEPOSIT","TRANSFER","BALANCE_CHECK","CREDIT_CARD_PAYMENT","ATM_WITHDRAWAL"])
print("46. Common transaction types:", rdd.intersection(rdd3).collect())
```

```python
print("47. Count common:", rdd.intersection(rdd3).count())
```

```python
print("48. Sorted common:", sorted(rdd.intersection(rdd3).collect()))
```

```python
print("49. Common containing CHECK:", rdd.intersection(rdd3).filter(lambda x: "CHECK" in x).collect())
```

```python
print("50. Common containing TRANSFER:", rdd.intersection(rdd3).filter(lambda x: "TRANSFER" in x).collect())
```

## FlatMap Exercises

```python
customer_comments = [
    "service was excellent",
    "mobile banking is easy",
    "transfer completed quickly",
    "deposit process was smooth",
    "excellent customer support",
    "loan payment completed successfully",
    "banking app is excellent",
    "service is fast and reliable"
]
comment_rdd = sc.parallelize(customer_comments)
words_rdd = comment_rdd.flatMap(lambda x: x.split())
print("66. All words:", words_rdd.collect())
```
