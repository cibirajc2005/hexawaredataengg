```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("OnlineLearning").getOrCreate()
sc = spark.sparkContext

activities = [
    "LOGIN","COURSE_VIEW","LOGIN","VIDEO_PLAY","QUIZ_START",
    "VIDEO_PLAY","LOGIN","COURSE_VIEW","ASSIGNMENT_SUBMIT","VIDEO_PLAY",
    "QUIZ_START","QUIZ_COMPLETE","LOGIN","VIDEO_PLAY","COURSE_VIEW",
    "LOGIN","VIDEO_PLAY","VIDEO_PLAY","QUIZ_START","LOGIN",
    "COURSE_VIEW","COURSE_VIEW","ASSIGNMENT_SUBMIT","LOGIN","VIDEO_PLAY",
    "QUIZ_COMPLETE","LOGIN","COURSE_VIEW","VIDEO_PLAY","VIDEO_PLAY",
    "LOGIN","COURSE_VIEW","QUIZ_START","ASSIGNMENT_SUBMIT","LOGIN",
    "VIDEO_PLAY","LOGIN","COURSE_VIEW","QUIZ_COMPLETE","VIDEO_PLAY",
    "LOGIN","COURSE_VIEW","VIDEO_PLAY","LOGIN","QUIZ_START",
    "COURSE_VIEW","VIDEO_PLAY","ASSIGNMENT_SUBMIT","LOGIN","VIDEO_PLAY"
]
rdd = sc.parallelize(activities)
print("RDD created successfully")
```

## CountByValue Exercises

```python
freq = rdd.map(lambda x: (x,1)).reduceByKey(lambda a,b: a+b)
print("36. Occurrences:", freq.collect())
```

```python
print("37. Most frequent:", freq.max(lambda x: x[1]))
```

```python
print("38. Least frequent:", freq.min(lambda x: x[1]))
```

```python
print("39. Frequencies sorted by count:")
[print(f"  {t}: {c}") for t,c in freq.sortBy(lambda x: -x[1]).collect()]
```

```python
print("40. Occurring more than 5 times:", freq.filter(lambda x: x[1] > 5).collect())
```

```python
print("45. New activities from rdd2:", rdd2.subtract(rdd).distinct().collect())
```

## FlatMap Exercises

```python
courses = [
    "Python SQL Spark",
    "Azure Databricks Spark",
    "Python Pandas NumPy",
    "SQL Spark DeltaLake",
    "Azure Python SQL"
]
course_rdd = sc.parallelize(courses)
words_rdd = course_rdd.flatMap(lambda x: x.split())
print("61. All words:", words_rdd.collect())
```

```python
print("62. Total words:", words_rdd.count())
```

```python
print("63. Distinct words:", words_rdd.distinct().collect())
```

```python
print("64. Sorted alphabetically:", sorted(words_rdd.distinct().collect()))
```

```python
word_freq = words_rdd.map(lambda x: (x,1)).reduceByKey(lambda a,b: a+b)
print("65. Word frequencies:", word_freq.collect())
```

```python
print("66. Words containing Python:", words_rdd.filter(lambda x: "Python" in x).collect())
```

```python
print("67. Words containing Spark:", words_rdd.filter(lambda x: "Spark" in x).collect())
```

```python
print("68. Uppercase words:", words_rdd.map(lambda x: x.upper()).collect())
```

```python
print("69. Longest word:", words_rdd.max(key=len))
```

```python
print("70. Shortest word:", words_rdd.min(key=len))
```

## Real RDD Analytics

```python
print("71. Most common activity:", freq.max(lambda x: x[1]))
```

```python
print("72. Least common activity:", freq.min(lambda x: x[1]))
```

```python
total = rdd.count()
login_count = rdd.filter(lambda x: x == "LOGIN").count()
print(f"73. LOGIN percentage: {login_count/total*100:.2f}%")
```

```python
vp_count = rdd.filter(lambda x: x == "VIDEO_PLAY").count()
print(f"74. VIDEO_PLAY percentage: {vp_count/total*100:.2f}%")
```

```python
print("75. Activity Report:")
print("="*40)
for t, c in freq.sortBy(lambda x: -x[1]).collect():
    print(f"  {t:<25} {c:>3} ({c/total*100:.1f}%)")
print("="*40)
```

```python
print("76. Unique Activity Report:")
[print(f"  {a}") for a in sorted(rdd.distinct().collect())]
```

```python
print("77. Frequency Report:")
[print(f"  {t}: {c}") for t,c in freq.sortBy(lambda x: -x[1]).collect()]
```

```python
print("="*55)
print("80. FINAL LEARNING PLATFORM USAGE SUMMARY")
print("="*55)
print(f"  Total Activity Records : {total}")
print(f"  Unique Activity Types  : {rdd.distinct().count()}")
print(f"\n  Activity Breakdown:")
for t, c in freq.sortBy(lambda x: -x[1]).collect():
    bar = "#" * c
    print(f"    {t:<25} {c:>3}  {bar}")
print(f"\n  Most Common  : {freq.max(lambda x: x[1])[0]}")
print(f"  Least Common : {freq.min(lambda x: x[1])[0]}")
print(f"  LOGIN %      : {rdd.filter(lambda x: x=='LOGIN').count()/total*100:.1f}%")
print(f"  VIDEO_PLAY % : {rdd.filter(lambda x: x=='VIDEO_PLAY').count()/total*100:.1f}%")
print("="*55)
```
